# 參數(parameter)，引數(argument)->真正的值

```python
def greet(name):
    print(f"哈囉,{name}!")

greet("小明")
```

### 引數位置{Position Argument}

```python
def describe_pet(animal, name):
    """顯示寵物資訊"""
    print(f"\n我有一隻{animal}.")
    print(f"我的{animal}叫做{name.title()}.")
    print(f"我養了一隻{animal}，它的名字叫{name}。")

#呼叫
describe_pet("小花", "貓")
describe_pet("貓", "小花")
```

### 引數名稱(Keyword Argument)

```python
def describe_pet(animal, name):
    print(f"我有一隻{animal}，它的名字叫{name}。")

describe_pet(animal="貓", name="小花")
describe_pet(name="小花", animal="貓")
```
